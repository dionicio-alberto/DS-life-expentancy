```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Datos
x = np.array([40, 34, 32, 45, 67, 53])
y = np.array([26, 23, 25, 27, 32, 29])

# Ajuste de regresión lineal
slope, intercept, r_value, p_value, std_err = linregress(x, y)

# Función de predicción
def predict(x_values):
    return slope * np.array(x_values) + intercept

# Valores predichos de y
y_pred = predict(x)

# Visualización
plt.scatter(x, y, color='blue', label='Datos originales')
plt.plot(x, y_pred, color='red', label=f'Regresión lineal: y = {slope:.2f}x + {intercept:.2f}')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()

# Mostrar ecuación y R²
print(f"Ecuación de la recta: y = {slope:.2f}x + {intercept:.2f}")
print(f"Coeficiente de correlación (R²): {r_value**2:.4f}")


```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 🔹 Generamos datos ficticios (tamaño en m² y precio en miles de dólares)
np.random.seed(42)
metros_cuadrados = np.random.randint(50, 200, 50)  # Tamaño entre 50 y 200 m²
precio_real = 100 + metros_cuadrados * 3 + np.random.normal(0, 15, 50)  # Relación lineal con ruido

# 🔹 Dividimos en conjunto de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(metros_cuadrados.reshape(-1,1), precio_real, test_size=0.2, random_state=42)

# 🔹 Entrenamos el modelo de regresión lineal
modelo = LinearRegression()
modelo.fit(X_train, y_train)

# 🔹 Predicciones
y_pred = modelo.predict(X_test)

# 🔹 Calculamos las métricas de error
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

# 🔹 Mostramos los resultados
print(f"🔹 MAE: {mae:.2f} (Error medio absoluto)")
print(f"🔹 MSE: {mse:.2f} (Error cuadrático medio)")
print(f"🔹 RMSE: {rmse:.2f} (Raíz del error cuadrático medio)")

# 🔹 Gráfica de los resultados
plt.scatter(metros_cuadrados, precio_real, label="Datos reales")
plt.plot(X_test, y_pred, color="red", label="Regresión lineal")
plt.xlabel("Metros cuadrados")
plt.ylabel("Precio (miles de dólares)")
plt.legend()
plt.title("Regresión Lineal: Tamaño de Casa vs Precio")
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Cargar datos de ejemplo (Iris dataset)
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=42)

# Crear el modelo de árbol de decisión con criterio Gini
clf = DecisionTreeClassifier(criterion="gini", max_depth=3, random_state=42)
clf.fit(X_train, y_train)

# Predicciones y evaluación
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print(f"Precisión del modelo: {accuracy:.2f}")
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Cargar datos (Iris dataset)
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=42)

# Crear y entrenar el árbol de decisión
clf = DecisionTreeClassifier(criterion="gini", max_depth=3, random_state=42)
clf.fit(X_train, y_train)

# Graficar el árbol
plt.figure(figsize=(12, 8))
plot_tree(clf, feature_names=data.feature_names, class_names=data.target_names, filled=True)
plt.show()
```
